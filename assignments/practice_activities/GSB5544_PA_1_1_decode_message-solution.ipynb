```python
step_1_ts = top_secret + 14
step_2_ts = (step_1_ts * 18) - 257
step_3_ts = np.exp(step_2_ts)
step_4_ts = step_3_ts**2
np.sum([step_4_ts < 17])
```

```python
step_5_ts = step_4_ts.reshape(-1,5)
evens = np.arange(2, 383, 2)
```

```python
# Step 5: reshape column-wise into 5 columns
step_5_ts = step_4_ts.reshape(-1, 5, order="F")
## Q1: Why is it important to do order equal "F"?
## key function: reshape


# Step 6: even numbers from 2 through 382
evens = np.arange(2, 383, 2)
## key function: arrange

# Step 7: subtract evens from first column
step_7_ts = step_5_ts.copy()
step_7_ts[:, 0] = step_7_ts[:, 0] - evens
## key function: copy
## key concept: {I had to slow this down. This is confusing.}

# Step 8: subtract 100 from rows 18-24 of the 3rd column
step_8_ts = step_7_ts.copy()
step_8_ts[17:24, 2] = step_8_ts[17:24, 2] - 100
## key function: copy
## key concept: index starts at 0, important for range and particular column

# Step 9: multiply all numbers in columns 4 and 5 by 2
step_9_ts = step_8_ts.copy()
step_9_ts[:, 3:5] = step_9_ts[:, 3:5] * 2
## key function: copy
## key concept: {I had to slow this down. This is confusing.}

# Step 10: turn matrix back into a 1D array, column-wise
step_10_ts = step_9_ts.flatten(order="F")
## Q2: Why is it important to do order equal "F"?
## key function: flatten

# Checkpoint: check that all values at index 500 and beyond are below 100
np.all(step_10_ts[500:] < 100)
## key concept: {I had to slow this down. This is confusing.}

# Step 11: square root of numbers in indices 37 through 464
step_11_ts = step_10_ts.copy()
step_11_ts[37:465] = np.sqrt(step_11_ts[37:465])
## key function: copy
## key concept: {I had to slow this down. This is confusing.}

# Step 12: round all numbers to nearest whole number
step_12_ts = np.round(step_11_ts)
## key function: round

# Step 13: replace every 39 with 20
step_13_ts = step_12_ts.copy()
step_13_ts[step_13_ts == 39] = 20
## key function: copy
## key concept: {I had to slow this down. This is confusing.}


# Checkpoint: count how many numbers are even
np.sum(step_13_ts % 2 == 0)
## key function: sum
## key concept: %s, == and
```
