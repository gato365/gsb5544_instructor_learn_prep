## Part 1 — One string: the core methods

Plain Python, one string at a time: `len`, indexing and slicing, `lower` / `upper`, `strip`, `replace`,
`split`, `join`, `startswith` / `endswith` — and the habit of **reassigning** the result.

### Solution 1 — `len`

```python
s = "  Data Science  "
len(s)
```

**Why it works.** `len` is a built-in *function* (not a method), so the string goes inside the parentheses. It counts every
character — the 2 leading spaces, the 12 characters of `Data Science` (the inner space included), and the 2 trailing spaces.

### Solution 2 — Indexing

```python
word = "regex"
(word[0], word[-1])
```

**Why it works.** Positions start at `0`, and negative positions count from the end, so `-1` is always the last character
whatever the length.

### Solution 3 — Slicing from the start

```python
title = "vaudevillian"
title[:5]
```

**Why it works.** `[:5]` means "from the start up to, but not including, position 5" — positions 0, 1, 2, 3, 4. This is how you
truncate a word to a maximum length (PA 4.2, step 3).

### Solution 4 — Slicing out of the middle

```python
code = "INV-2024-00931"
code[4:8]
```

**Why it works.** `INV-` occupies positions 0–3, so the year starts at position 4; the slice stops *before* position 8, giving
positions 4, 5, 6, 7.

### Solution 5 — `lower` and `upper`

```python
name = "McDonald"
(name.lower(), name.upper())
```

**Why it works.** Both are *methods*: they come after the string and a dot, and need the empty parentheses to actually run.

### Solution 6 — `strip`

```python
raw = "   hello world   "
raw.strip()
```

**Why it works.** `strip()` removes whitespace from the **two ends only**; the space between `hello` and `world` is in the middle,
so it stays.

### Solution 7 — `strip` changes the length

```python
raw = "   hello world   "
(len(raw), len(raw.strip()))
```

**Why it works.** 17 characters with padding, 11 without: three spaces came off each end. Methods can be used inside a function
call — `raw.strip()` produces a new string, and `len` measures that.

### Solution 8 — `replace`

```python
text = "I like cats. cats are great."
text.replace("cats", "dogs")
```

**Why it works.** `replace(old, new)` swaps **every** occurrence of the exact text `old`, not just the first one.

### Solution 9 — `replace` to delete

```python
phone = "805-555-0199"
phone.replace("-", "")
```

**Why it works.** Replacing with the empty string `""` is how you delete text. The result is still a string, not a number.

### Solution 10 — `split`

```python
line = "red,green,blue"
line.split(",")
```

**Why it works.** `split(sep)` cuts the string wherever `sep` occurs and returns a **list** of the pieces; the separators
themselves are thrown away.

### Solution 11 — `split` to count words

```python
sentence = "the quick brown fox"
len(sentence.split(" "))
```

**Why it works.** Splitting on `" "` gives `['the', 'quick', 'brown', 'fox']`, and `len` of a list is its number of elements.

### Solution 12 — `join`

```python
pieces = ["2026", "09", "17"]
"-".join(pieces)
```

**Why it works.** `join` is a method of the **separator**, and the list goes in the parentheses: `"-".join(pieces)`. It is the
reverse of `split`.

### Solution 13 — `split`, then `join`

```python
text = "week four notes"
"_".join(text.split(" "))
```

**Why it works.** `split(" ")` gives `['week', 'four', 'notes']`; `"_".join(...)` stitches the list back together with the new
separator. (`text.replace(" ", "_")` gets the same answer — two routes, one result.)

### Solution 14 — `startswith` and `endswith`

```python
fname = "report_final.csv"
(fname.startswith("report"), fname.endswith(".csv"))
```

**Why it works.** Both return `True` or `False`. They test exact text at one end of the string — no patterns involved.

### Solution 15 — Strings are immutable

```python
s = "  padded  "
s = s.strip()
len(s)
```

**Why it works.** No string method changes the original; each returns a **new** string. Writing `s.strip()` alone computes the
answer and discards it — you must reassign with `s = s.strip()`. Without the fix, `len(s)` would still be 10.

### Solution 16 — Chaining methods

```python
messy = "   ThE DuDe AbIdEs.  "
messy.strip().lower()
```

**Why it works.** Each method returns a string, so another method can follow immediately. Chains run **left to right**: strip
first, then lowercase the stripped result.

### Solution 17 — A longer chain

```python
messy = "   ThE DuDe AbIdEs.  "
messy.strip().split(" ")[-1].upper()
```

**Why it works.** `strip()` first (otherwise the trailing spaces would create empty pieces), `split(" ")` makes a list, `[-1]`
takes its last element, and `.upper()` works on that string. The full stop is part of the "word" — `split`
knows nothing about punctuation.

## Part 2 — Regular expressions: finding with `re.findall`

`re.findall(pattern, s)` returns a list of every match. Write patterns as **raw strings**, `r"..."`.
The pieces: literal text, sets `[...]`, negated sets `[^...]`, `\w`, `\s`, `.`, the quantifiers `+`, `*`, `{n}`,
and the anchors `^` and `$`.

```python
import re
```

### Solution 18 — A literal pattern

```python
text = "the cat sat on the mat with the hat"
re.findall(r"the", text)
```

**Why it works.** Ordinary letters in a pattern match themselves. `findall` returns one list element per match, in order.

### Solution 19 — A pattern matches anywhere

```python
text = "other theory, the end"
len(re.findall(r"the", text))
```

**Why it works.** Three: inside `o-the-r`, inside `the-ory`, and the standalone `the`. A regex has no idea what a *word* is — a
literal pattern matches wherever those characters appear.

### Solution 20 — A set of characters

```python
text = "regular expression"
re.findall(r"[aeiou]", text)
```

**Why it works.** Square brackets mean "any **one** character from this set". Each vowel is its own match, so the list has
seven single-character elements.

### Solution 21 — A set of digits

```python
text = "Room 4B, floor 12"
re.findall(r"[0123456789]", text)
```

**Why it works.** The set can hold any characters, here the ten digits. `12` comes back as `'1'` and `'2'` because a set matches
**one** character. (Shorthand you will see elsewhere: `[0-9]` is the same set.)

### Solution 22 — A negated set

```python
re.findall(r"[^aeiou]", "banana")
```

**Why it works.** A `^` as the **first thing inside the brackets** flips the set: "any one character *except* these".

### Solution 23 — The punctuation class

```python
text = "Wait... what?! No way; really."
re.findall(r"[^\w\s]", text)
```

**Why it works.** `\w` is a word character (letter, digit, underscore) and `\s` is whitespace. `[^\w\s]` is "not a word character
and not whitespace" — which leaves punctuation. The backslashes are why the pattern is a raw string: in
`r"..."` Python passes `\w` to the regex engine untouched.

### Solution 24 — `\s` — whitespace

```python
text = "a b\tc\nd"
len(re.findall(r"\s", text))
```

**Why it works.** `\s` matches any single whitespace character: the space, the tab `\t`, and the newline `\n`. This is how the
PA's "how many characters are white space?" question is answered.

### Solution 25 — `\w` — word characters

```python
text = "hi there!"
len(re.findall(r"\w", text))
```

**Why it works.** `\w` matches letters, digits, and the underscore — `h i t h e r e` is seven. The space and the `!` are not
word characters.

### Solution 26 — `.` — any character

```python
text = "cat cot cut c_t"
re.findall(r"c.t", text)
```

**Why it works.** The dot is a wildcard for exactly one character of any kind, so `a`, `o`, `u`, and `_` all qualify.

### Solution 27 — `+` — one or more

```python
text = "go goo gooooal g"
re.findall(r"go+", text)
```

**Why it works.** `+` applies to the single character before it: `o+` is "one or more `o`". The `+` is greedy — it takes as many
`o`s as it can. The lone `g` at the end has no `o`, so it does not match.

### Solution 28 — `*` — zero or more

```python
text = "go goo gooooal g,"
re.findall(r"go*", text)
```

**Why it works.** `o*` also accepts *no* `o` at all, so the lone `g` now matches. `+` = at least one; `*` = optional and repeatable.

### Solution 29 — `\w+` — whole words

```python
text = "Voila! In view, a humble veteran."
re.findall(r"\w+", text)
```

**Why it works.** `\w+` is "a run of one or more word characters". A run ends at the first space or punctuation mark, so each
match is one word.

### Solution 30 — `{n}` — exactly n

```python
text = "meet me at the beet tree"
re.findall(r"e{2}", text)
```

**Why it works.** `e{2}` is "the character `e`, exactly twice" — the same as writing `ee`. It matches in `meet`, `beet`, and
`tree`; the single `e`s in `me` and `the` do not qualify.

### Solution 31 — A literal followed by a quantifier

```python
groans = "ugh ughh ughhhh! Ugh"
re.findall(r"ugh+", groans)
```

**Why it works.** `ug` is literal; `h+` is one or more `h`. The `+` belongs only to the `h`, not to the whole word. The last
groan, `Ugh`, is missed — see the next exercise.

### Solution 32 — Regex is case-sensitive

```python
groans = "ugh ughh ughhhh! Ugh"
re.findall(r"[Uu]gh+", groans)
```

**Why it works.** A lowercase `u` in a pattern only matches a lowercase `u`. The set `[Uu]` accepts either case for that one
position; the rest of the pattern is unchanged.

### Solution 33 — `^` — at the start

```python
text = "the end of the road"
re.findall(r"^the", text)
```

**Why it works.** Outside brackets, `^` is an **anchor**: it matches a position (the start of the string), not a character. The
second `the` is in the middle, so only one match comes back. (Inside brackets, `^` means "not" — same symbol,
different job.)

### Solution 34 — `$` — at the end

```python
text = "singing and dancing"
re.findall(r"ing$", text)
```

**Why it works.** `$` anchors to the end of the string. `singing` contains `ing` twice, but neither is at the very end; only the
`ing` of `dancing` is.

### Solution 35 — Escaping a special character

```python
text = "v1.2 and v1x2"
re.findall(r"1\.2", text)
```

**Why it works.** `.` normally means "any character", so `1.2` would match both `1.2` and `1x2`. A backslash turns a special
character back into a literal: `\.` is "an actual full stop". In a raw string you type one backslash; in an
ordinary string you would need `"1\\.2"`.

### Solution 36 — Escaping `$`

```python
text = "cost: $5 and $12"
re.findall(r"\$\w+", text)
```

**Why it works.** `$` is the end-of-string anchor, so to match a real dollar sign it must be escaped as `\$`. Then `\w+` takes
the digits that follow. The same rule applies to the other special characters: `^ $ . + * [ ] { }`.

### Solution 37 — Combining a quantifier and a class

```python
text = "rough day, ugh, enough!"
re.findall(r"ugh+[^\w\s]", text)
```

**Why it works.** `ugh+` then exactly one punctuation character. The `ugh` in `rough` is followed by a space, so it is left
alone — but the one in `enough!` is followed by `!` and **is** caught. Testing with `findall` first shows you
that this pattern would damage `enough` before you ever run a replacement.

## Part 3 — Regular expressions: replacing with `re.sub`

`re.sub(pattern, replacement, s)` replaces every match. Same patterns as Part 2 — now they change the string.

### Solution 38 — `re.sub` with a literal

```python
date = "2026-09-17"
re.sub(r"-", "/", date)
```

**Why it works.** The argument order is pattern, replacement, string. With a purely literal pattern this does the same job as
`date.replace("-", "/")`.

### Solution 39 — Delete all punctuation

```python
text = "Hello, world! Ready?"
re.sub(r"[^\w\s]", "", text)
```

**Why it works.** `[^\w\s]` matches one punctuation character at a time; replacing each with `""` deletes it. The spaces survive
because `\s` is excluded from the match.

### Solution 40 — Collapse repeated spaces

```python
text = "too    many   spaces"
re.sub(r"\s+", " ", text)
```

**Why it works.** `\s+` matches a whole run of whitespace as one match, and the run is replaced by one space. With `\s` alone
(no `+`) each space would be replaced by a space and nothing would change.

### Solution 41 — Delete the groans

```python
text = "Fineugh! I will go.Ughhh? Ok"
re.sub(r"[Uu]gh+[^\w\s]", "", text)
```

**Why it works.** `[Uu]` = either case, `gh+` = `g` and one or more `h`, `[^\w\s]` = the punctuation mark after it. The full stop
after `go` is kept because it comes *before* the groan and is not part of any match.

### Solution 42 — Anchored replacement

```python
word = "kicking"
re.sub(r"^k", "v", word)
```

**Why it works.** `^k` matches a `k` only at position 0, so the second `k` is untouched. `word.replace("k", "v")` would have
produced `vicving`.

### Solution 43 — Anchored replacement at the end

```python
word = "bubblb"
re.sub(r"b$", "y", word)
```

**Why it works.** `b$` is "a `b` that is the last character". The three earlier `b`s are not at the end, so they stay.

## Part 4 — A whole column at once: the `.str` accessor

The same tools applied to every element of a pandas Series: `.str.strip`, `.str.len`, `.str.upper`,
`.str.slice`, `.str.contains`, `.str.replace` (remember `regex=True`), and `.apply` + `lambda` for the rest.
Run the set-up cells as they appear; later exercises use `words` and `clean`.

```python
import pandas as pd

words = pd.Series(["  Koila! ", "kiew,", "  humble  ", "kezeranugh!", "bozh", "kicarb"])
words
```

### Solution 44 — `.str.strip`

```python
words.str.strip()
```

**Why it works.** `.str` is the doorway from a Series to the string methods: `words.str.strip()` runs `strip()` on each element
and returns a new Series. `words.strip()` without `.str` is an error.

```python
clean = words.str.strip()      # set-up: run this whether or not you finished the previous exercise
```

### Solution 45 — `.str.len`

```python
clean.str.len()
```

**Why it works.** There is no `len(...)` for "each element", so pandas supplies `.str.len()`. `len(clean)` would instead return
6 — the number of *elements* in the Series.

### Solution 46 — Total characters

```python
clean.str.len().sum()
```

**Why it works.** `.str.len()` gives a Series of numbers, and `.sum()` adds them — the pattern for PA 4.2's "how many characters
are in the scrambled message?"

### Solution 47 — `.str.upper`

```python
clean.str.upper()
```

**Why it works.** Same idea as `.str.strip()`: the plain string method `upper`, applied element by element.

### Solution 48 — `.str.slice`

```python
clean.str.slice(0, 4)
```

**Why it works.** `.str.slice(0, 4)` is the Series version of `word[0:4]`. Words that are already short (`bozh`) are returned
unchanged.

### Solution 49 — `.str.contains` with an anchor

```python
clean.str.contains(r"^k")
```

**Why it works.** `.str.contains` treats its argument as a regex by default, so `^k` means "starts with `k`". `Koila!` is
`False` because the pattern is case-sensitive.

### Solution 50 — `case=False`, and filtering

```python
clean[clean.str.contains(r"^k", case=False)]
```

**Why it works.** `case=False` makes the match ignore capitalisation. The `True`/`False` Series goes inside `clean[...]` to keep
only the rows where it is `True` — ordinary pandas filtering.

### Solution 51 — Words that end in punctuation

```python
clean[clean.str.contains(r"[^\w\s]$")]
```

**Why it works.** `[^\w\s]` = one punctuation character, `$` = at the end. Both pieces are from Part 2; here they run on every
element.

### Solution 52 — `.str.replace` with literal text

```python
clean.str.replace("z", "t")
```

**Why it works.** For plain text, `.str.replace(old, new)` behaves like the string method, on every element. No `regex=True`
is needed because `z` is just a letter.

### Solution 53 — `.str.replace` with a pattern

```python
clean.str.replace(r"^k", "v", regex=True)
```

**Why it works.** In `.str.replace` regex is **off** by default, so a pattern needs `regex=True`. `Koila!` still starts with a
capital `K` and is untouched — a second replacement with `^K` → `V` would handle it.

### Solution 54 — The silent bug

```python
clean.str.replace("^k", "v").equals(clean)
```

**Why it works.** `True` — nothing changed. Without `regex=True`, pandas searches for the two literal characters `^k`, finds
none, and "succeeds" at replacing zero occurrences. No error message, wrong answer.

### Solution 55 — Delete the groans from a column

```python
clean.str.replace(r"ugh+[^\w\s]", "", regex=True)
```

**Why it works.** The same pattern as in Part 3, now on a column. Only `kezeranugh!` contains a groan followed by punctuation;
it becomes `kezeran`. The `!` of `Koila!` is not preceded by `ugh`, so it stays.

### Solution 56 — `.apply` + `lambda`

```python
clean.apply(lambda w: re.findall(r"[^\w\s]", w))
```

**Why it works.** `.apply` hands each element to the function; `lambda w: ...` is a one-line function whose argument `w` is one
word. Every element of the result is a list — empty when the word has no punctuation.

### Solution 57 — `.apply` to count

```python
clean.apply(lambda w: len(re.findall(r"[aeiou]", w)))
```

**Why it works.** Inside the `lambda`, `re.findall` returns the list of vowels in one word and `len` counts them. Anything you
can do to one string, `.apply` + `lambda` does to the whole Series.

### Solution 58 — Preview for PA 4.2 — `.str.count`

```python
words.str.count(r"\s").sum()
```

**Why it works.** `.str.count(r"\s")` does what the previous exercise did with `.apply`, in one step: matches of `\s` per element
(3, 0, 4, 0, 0, 0). `.sum()` totals them. This answers "how many characters are white space?"

### Solution 59 — Preview for PA 4.2 — keeping part of the match with `\1`

```python
endings = pd.Series(["kicarb", "kanizb,", "bb.", "bubble"])
endings.str.replace(r"b([^\w\s]*)$", r"y\1", regex=True)
```

**Why it works.** `b` = the letter to change; `([^\w\s]*)` = zero or more punctuation marks, **captured** by the parentheses;
`$` = end of the word. In the replacement, `\1` means "whatever the first pair of parentheses matched", so the
comma or full stop is preserved. `*` (not `+`) lets the pattern also match when there is no punctuation at all.
The replacement is a raw string too, so that `\1` reaches pandas intact.

### Solution 60 — Series to one string

```python
" ".join(clean)
```

**Why it works.** `join` accepts any collection of strings, including a Series. The separator comes first, as always:
`" ".join(clean)`.

## Part 5 — Mixed exercises: you choose the tools

The blanks are larger now — whole expressions. Decide which methods and patterns the job needs.

### Solution 61 — Tidy product codes

```python
codes = pd.Series(["  ab-101 ", "AB-102", " ab-103x "])
codes.str.strip().str.upper().str.slice(0, 6)
```

**Why it works.** Three `.str` methods chained. Order matters: strip **before** slicing, otherwise the leading spaces would use
up some of the six characters.

### Solution 62 — Normalise a messy string

```python
messy = "   Too   MANY    Spaces  "
re.sub(r"\s+", " ", messy.strip()).lower()
```

**Why it works.** `strip()` handles the ends, `re.sub(r"\s+", " ", ...)` turns each inner run of whitespace into one space, and
`.lower()` finishes. `re.sub` returns a string, so a string method can be chained onto it.

### Solution 63 — Extract hashtags

```python
post = "Loving #python and #regex, not #Mondays!"
re.findall(r"#\w+", post)
```

**Why it works.** `#` is not special in a regex, so it needs no escaping; `\w+` takes the letters after it and stops at the comma,
space, or `!`.

### Solution 64 — Count the -ing words

```python
sentence = "Singing, dancing and acting are tiring things."
pd.Series(sentence.split(" ")).str.contains(r"ing[^\w\s]*$").sum()
```

**Why it works.** Split into a Series of words, then test each: `ing`, then zero or more punctuation marks (`[^\w\s]*`), then the
end (`$`). Without the punctuation part, `Singing,` would be missed. `things.` ends in `ings.`, not `ing`, so it does not
count. Summing `True`/`False` counts the `True`s.

```python
secret = pd.Series(["  Xhe ", "ughh!qreasure", "  is", "buriedugh?", "  undez  ", "xhe", "flooz."])
```

### Solution 65 — Warm-up on a scrambled message

```python
(secret.str.len().sum(), secret.str.count(r"\s").sum())
```

**Why it works.** `.str.len().sum()` totals the lengths; `.str.count(r"\s").sum()` totals the whitespace characters (3 + 0 + 2 +
0 + 4 + 0 + 0).

### Solution 66 — The longest word, in capitals

```python
stripped = secret.str.strip()
stripped[stripped.str.len() == stripped.str.len().max()].str.upper()
```

**Why it works.** Compare each length with the maximum length to get a `True`/`False` filter, keep the matching row, and
capitalise it. Strip first — otherwise padding would count towards the length.

### Solution 67 — Decode the message

```python
decoded = (secret
           .str.strip()
           .str.replace(r"ugh+[^\w\s]", "", regex=True)
           .str.replace("q", "t")
           .str.replace(r"^x", "t", regex=True)
           .str.replace(r"^X", "T", regex=True)
           .str.replace(r"z([^\w\s]*)$", r"r\1", regex=True))
" ".join(decoded)
```

**Why it works.** The PA 4.2 pipeline in miniature. Step 2 uses `ugh+[^\w\s]`; step 3 is plain text, so no `regex=True`; step 4
needs two anchored replacements because regex is case-sensitive; step 5 captures the optional punctuation with
`([^\w\s]*)` and restores it with `\1`, which is what turns `flooz.` into `floor.` rather than leaving it alone.

### Solution 68 — Pieces of words

```python
found = decoded.apply(lambda w: re.findall(r"t\w*e", w))
[piece for pieces in found for piece in pieces]
```

**Why it works.** `t`, then zero or more word characters (`\w*`), then `e`. Because `*` is greedy, `treasure` is matched whole
rather than stopping at the first `e`. `The` is not matched: its `T` is a capital. The last line flattens the
Series of lists into one list.
